# 📊 **Customer Investment Needs & Recommendation System**

## **Project Overview**
This project aims to build a **data-driven recommendation system** for a wealth management context.

Using real-world customer data, we seek to **infer investment needs** and translate them into **personalized product recommendations**, also known as the **Next Best Action**.

The workflow combines:
- **Machine Learning** for need identification
- **Financial reasoning** for product suitability
- **Data Science techniques** for scalable decision-making

---

## **🎯 Objectives**
The project is structured around two main goals:

1. **Estimate Customer Investment Needs**
   - Predict whether a client has a **high propensity** for:
     - **Accumulation investing** (regular contributions over time)
     - **Income investing** (lump-sum / income-generating investments)

2. **Build a Recommendation Engine**
   - Suggest the **most suitable financial product** for each customer based on:
     - Identified investment need
     - Risk compatibility

---

## **💡 Motivation**
Understanding customer needs is central to modern finance:

- **Personalization:** Deliver tailored investment strategies
- **Client Advisory (KYC):** Improve decision-making quality
- **Regulatory Compliance:** Ensure alignment with frameworks like **MiFID / IDD**
- **Automation:** Scale advisory services using data

This project can be interpreted as building a **data-driven financial advisor**.

---

## **🧠 Methodological Approach**
The solution follows a two-step pipeline:

1. **Classification Layer**
   - Learn customer behavior from historical data
   - Predict investment needs

2. **Recommendation Layer**
   - Map predicted needs to suitable financial products
   - Incorporate **risk alignment** for final recommendations
